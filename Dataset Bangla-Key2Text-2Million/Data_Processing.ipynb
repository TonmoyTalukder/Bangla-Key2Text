### Clean English Alphabet

```python
import pandas as pd
df  = pd.read_csv("/content/drive/MyDrive/Shibli Sir/Multi-Modal Research Team/Work 0: Keywords to Text in Bengali/DataV3/strkeysDfVtest0.csv")
df
```

```python
import csv
import re

# Open the CSV file and create a reader object
with open("/content/drive/MyDrive/Shibli Sir/Multi-Modal Research Team/Work 0: Keywords to Text in Bengali/DataV3/strkeysDfVtest0.csv", 'r') as file:
    reader = csv.reader(file)

    # Get the header row and find the index of the specific column
    header = next(reader)
    column_index = header.index('bn_Sentence')

    # Create a new list to hold the rows that don't contain English words
    clean_rows = []

    # Loop through each row in the file
    for row in reader:

        # Check if the specific column contains any English words
        if not re.search(r'[a-zA-Z0-9]', row[column_index]):

            # If there are no English words, add the row to the clean_rows list
            clean_rows.append(row)

    # Write the clean rows to a new CSV file
    with open('/content/drive/MyDrive/Shibli Sir/Multi-Modal Research Team/Work 0: Keywords to Text in Bengali/DataV3/strkeysDfVclean_file.csv', 'w', newline='') as clean_file:
        writer = csv.writer(clean_file)
        writer.writerow(header)
        writer.writerows(clean_rows)
```

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Shibli Sir/Multi-Modal Research Team/Work 0: Keywords to Text in Bengali/DataV3/strkeysDfV01.csv")
df = df.head(2000000)
df
```

```python
def find_words_number(sentence):
    words = sentence.split()
    return len(words)
```

```python
df['Words in KeyWords'] = df['bn_Keywords'].apply(find_words_number)
df['Words in Sentence'] = df['bn_Sentence'].apply(find_words_number)
```

```python
print(df['Words in Sentence'].min(), df['Words in Sentence'].max(), df['Words in Sentence'].mean())
```

```python
from itertools import combinations
from collections import Counter

def calculate_keyword_frequency(dataframe):
    keyword_frequency = Counter(df['bn_Keywords'])
    return keyword_frequency.most_common(5)

def perform_cooccurrence_analysis(dataframe):
    cooccurrence_matrix = Counter()
    for keywords in df['bn_Keywords']:
        keyword_combinations = combinations(keywords, 2)
        cooccurrence_matrix.update(keyword_combinations)
    return cooccurrence_matrix.most_common(5)

def calculate_keyword_diversity(dataframe):
    keyword_diversity = df['bn_Keywords'].apply(lambda x: len(set(x))).mean()
    return keyword_diversity

# Calculate keyword frequency
keyword_frequency = calculate_keyword_frequency(df)
print("Top 5 Keyword Frequency:")
for keyword, frequency in keyword_frequency:
    print(keyword, ":", frequency)
print()

# Perform co-occurrence analysis
cooccurrence_matrix = perform_cooccurrence_analysis(df)
print("Top 5 Co-occurrence Matrix:")
for (keyword1, keyword2), frequency in cooccurrence_matrix:
    print(keyword1, "-", keyword2, ":", frequency)
print()

# Calculate keyword diversity
keyword_diversity = calculate_keyword_diversity(df)
print("Keyword Diversity:", keyword_diversity)
```

```python
from collections import Counter

def calculate_keyword_distribution(df):
    keyword_frequency = Counter()
    for keywords in df['bn_Keywords']:
        keyword_frequency.update(keywords)
    return keyword_frequency.most_common(10)

def analyze_keyword_distribution(df):
    keyword_distribution = calculate_keyword_distribution(df)
    total_texts = len(df)

    for keyword, frequency in keyword_distribution:
        percentage = (frequency / total_texts) * 100
        print(f"Keyword: {keyword}, Frequency: {frequency}, Percentage: {percentage:.2f}%")


# Analyze top 5 keyword distribution across texts
print("Top 10 Keyword Distribution Across Texts:")
analyze_keyword_distribution(df)
```
