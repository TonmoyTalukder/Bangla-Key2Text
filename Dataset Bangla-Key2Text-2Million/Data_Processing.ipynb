## Preprocessing

```python
import pandas as pd

df = pd.read_csv("./Data/Training Data/split600K.csv")
df
```

### Clean Special Characters and Others

```python
import string
invalidcharacters= set(string.punctuation)

def invalidcharacter(key):
    if key in invalidcharacters:
        return True
    else:
        return False
```

```python
def cleanDataX(keywords):

  strs = ''
  Flag = 0
  for i in range(len(keywords)):

    if keywords[i] == '[':
      continue
    elif keywords[i] ==']':
      continue
    elif keywords[i] =='।': # dari
      continue
    elif keywords[i] =='‘': # 4278
      continue
    elif keywords[i] =='’': # 4278
      continue
    elif keywords[i] == chr(39): # '
      continue
    elif keywords[i] == chr(44): # ,
      continue
    # elif checkEnglish(keywords[i]) == True and invalidcharacter(keywords[i]) == True:
    #   continue
    elif invalidcharacter(keywords[i]) == True:
      continue
    else:
      strs = strs + keywords[i]

  # return ''.join((x for x in strs if not x.isdigit()))
  return ''.join(strs)

df['bn_Keywords'] = df['bn_Keywords'].apply(cleanDataX)

df.to_csv("./Data/Training Data/split600K.csv", index=False)
```

```python
import csv
import re

# Open the CSV file and create a reader object with 'utf-8' encoding
with open("./Data/Training Data/split600K.csv", 'r', encoding='utf-8') as file:
    reader = csv.reader(file)

    # Get the header row and find the index of the specific column
    header = next(reader)
    column_index = header.index('bn_Sentence')

    # Create a new list to hold the rows that don't contain English words
    clean_rows = []

    # Loop through each row in the file
    for row in reader:

        # Check if the specific column contains any English words
        if not re.search(r'[a-zA-Z0-9]', row[column_index]):

            # If there are no English words, add the row to the clean_rows list
            clean_rows.append(row)

    # Write the clean rows to a new CSV file
    with open('./Data/Training Data/split600K.csv', 'w', newline='', encoding='utf-8') as clean_file:
        writer = csv.writer(clean_file)
        writer.writerow(header)
        writer.writerows(clean_rows)
```
